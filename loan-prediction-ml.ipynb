# **First convert categorical data to numerical use Ordinal encoding**

```python
from sklearn.preprocessing import OrdinalEncoder

encode1=OrdinalEncoder(categories=[[" Graduate"," Not Graduate"]])
encode2=OrdinalEncoder(categories=[[" Yes"," No"]])
```

```python
df[["Encode_education"]]=encode1.fit_transform(df[[" education"]])
df[["Encode_employed"]]=encode2.fit_transform(df[[" self_employed"]])
```

```python
df.head()
```

```python
df.drop(columns=[" education"," self_employed"],inplace=True)
```

# **Remove categorical input part**

```python
df.head()
```

# **Split the data**

```python
from sklearn.model_selection import train_test_split

x_train,x_test,y_train,y_test=train_test_split(df.drop(columns=[" loan_status"]),df[" loan_status"],test_size=0.2,random_state=42)
```

```python
x_train.head()
```

# **Now, Convert categorical y_train column with label encoder**

```python
from sklearn.preprocessing import LabelEncoder
```

```python
le=LabelEncoder()
```

```python
y_train=le.fit_transform(y_train)
```

```python
y_test=le.transform(y_test)
```

```python
y_train
```

# **Import All features**

```python
from sklearn.pipeline import make_pipeline
from sklearn.feature_selection import SelectKBest,f_classif
from sklearn.model_selection import cross_val_score
from sklearn.metrics import accuracy_score
from sklearn.tree import DecisionTreeClassifier
```

```python
trf1=SelectKBest(score_func=f_classif,k=10)
```

```python
trf2=DecisionTreeClassifier()
```

```python
pipe=make_pipeline(trf1,trf2)
```

```python
pipe.fit(x_train,y_train)
```

```python
y_predict=pipe.predict(x_test)
```

```python
accuracy_score(y_predict,y_test)
```

# **Now Apply cross validation**

```python
cross_val_score(pipe,x_train,y_train,cv=5,scoring='precision').mean()
```

```python
cross_val_score(pipe,x_train,y_train,cv=5,scoring='recall').mean()
```

```python
cross_val_score(pipe,x_train,y_train,cv=5,scoring='f1').mean()
```

# **Use Random forest calssifier**

```python
from sklearn.ensemble import RandomForestClassifier
```

```python
trf3=RandomForestClassifier()
```

```python
pipe=make_pipeline(trf1,trf3)
pipe.fit(x_train,y_train)
y_predict=pipe.predict(x_test)
```

```python
accuracy_score(y_predict,y_test)
```

```python
cross_val_score(pipe,x_train,y_train,cv=5,scoring='precision').mean()
```
